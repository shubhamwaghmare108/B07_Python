```python
match = re.search(r"(?P<area>\d{3})-(?P<exchange>\d{3})-(?P<number>\d{4})", "555-123-4567")
print(match.group("area"))      # 555
print(match.groupdict())        # {'area': '555', 'exchange': '123', 'number': '4567'}
```

```python
match = re.search(r"(\d{3})-(\d{3})-(\d{4})", "555-123-4567")
print(match.group(0[0]))      # 555
print(match.groupdict())        # {'area': '555', 'exchange': '123', 'number': '4567'}
```

```python
match = re.search(r"(\d{3})-(\d{3})-(\d{4})", "555-123-4567")
print(match[0].group(0))   # 555
print(match.groupdict())        # {'area': '555', 'exchange': '123', 'number': '4567'}
```

```python
import re

text = "scwagh123@gmail.com asdf@yahoo.in"

print(re.findall(r"\w+@(gmail|yahoo|outlook)\.(in|com)", text))
```

```python
pattern = re.compile(r"((?:\w+(?:\.?\w?)*)@(?:gmail|yahoo|outlook)\.(?:in|com))")
```

```python
re.match(pattern,'scwagh123@gmail.com')
```

```python
re.search(pattern,'scwagh123@gmail.com')
```

```python
re.findall(pattern,'scwagh123@gmail.com shubham.c.waghmare@gmail.com  scertgd@gmail.com')
```
