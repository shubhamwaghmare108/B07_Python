```python
1. Shopping Bill Calculator
A customer buys products with different prices and quantities. 
Calculate the total bill. If the bill exceeds ₹5,000, apply a 10% discount.
```

```python
price = []
qt = []
total_bill = 0
q = False
discount = 0

while q == False:
    x = float(input("Enter unit price:\n"))
    y = float(input("Enter qt purchased:\n"))

    if x == 0  or y == 0:
        q = True
    else:
        price.append(x)
        qt.append(y)

    
for a,b in zip(price,qt):
    total_bill += a * b
if total_bill > 5000:
    discount = total_bill * 0.1
    net_amount = total_bill * 0.9
else:
    net_amount = total_bill

print("Bill Summary".center(30,"*"))
print("Total Bill",total_bill)
print("Discount",discount)
print("Amount to be Paid",net_amount)
print("End".center(30,"*"))
```

```python
while True:
    try:
        marks = input(
            "Enter marks for 5 subjects separated by commas:\n"
        ).split(",")

        if len(marks) != 5:
            print("Please enter marks for exactly 5 subjects.")
            continue

        final_marks = [float(mark.strip()) for mark in marks]

        if any(mark < 0 or mark > 100 for mark in final_marks):
            print("Marks must be between 0 and 100.")
            continue

        break

    except ValueError:
        print("Invalid input! Enter numeric marks only.")

# Calculate percentage
total_marks = sum(final_marks)
percentage = (total_marks / 500) * 100

# Determine result
if any(mark < 35 for mark in final_marks):
    result = "FAIL"
    grade = "F"
elif percentage >= 90:
    result = "PASS"
    grade = "A+"
elif percentage >= 80:
    result = "PASS"
    grade = "A"
elif percentage >= 70:
    result = "PASS"
    grade = "B"
elif percentage >= 60:
    result = "PASS"
    grade = "C"
elif percentage >= 35:
    result = "PASS"
    grade = "D"

# Display result
print("\n" + " STUDENT RESULT ".center(35, "*"))
print(f"Subject Marks : {final_marks}")
print(f"Total Marks   : {total_marks:.2f}/500")
print(f"Percentage    : {percentage:.2f}%")
print(f"Result        : {result}")
print(f"Grade         : {grade}")
print("*" * 35)
```

```python
A person enters expenses until they type done. 
Calculate total spending, average expense, highest expense,
and the number of expenses above ₹500.
```

```python
expenses = []

var = ""
while var != "done":
    var = input("Enter your expence and if done type 'done' to exit")
    if var != 'done':
        expenses.append(float(var))
total_spending = sum(expenses)
average_expense = total_spending / len(expenses)
highest_expense = max(expenses)
exp_above_500 = 0
for i in expenses:
    if i > 500:
        exp_above_500 += 1
```

```python
print(total_spending)
print(average_expense)
print(highest_expense)
print(exp_above_500)
```
