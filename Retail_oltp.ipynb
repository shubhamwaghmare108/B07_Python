```python
import os

# List of DataFrame names
csv_list = ['brands','categories','customers','inventory','lookup_channel','lookup_gender',
            'lookup_order_status','lookup_payment_mode','product_suppliers','products',
            'sales_order_items','sales_orders','stores','suppliers']

# Create a folder to store CSVs
folder = "retail_csvs"
os.makedirs(folder, exist_ok=True)

# Loop through each DataFrame name and save to CSV
for name in csv_list:
    # Use globals() to fetch the DataFrame object by its variable name
    df = globals()[name]
    df.to_csv(os.path.join(folder, f"{name}.csv"), index=False)
    print(f"Saved {name}.csv in {folder}/")
```
