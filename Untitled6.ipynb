```python
text = "I love Python"
```

```python
def counter(text):
    result = {}
    var = text
    for i in var:
        try:
            result[i.lower()] += 1  
        except KeyError:
            result[i.lower()] = 1
    return result
```

```python
result = counter(text)
```
