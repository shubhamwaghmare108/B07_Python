```python
• strftime() : Formats a time object into a string.
• strptime(): Parses a string into a time object.
```

```python
now = datetime.datetime.now()

# Convert Datetime object to String (Format: YYYY-MM-DD HH:MM:SS)
string_date = now.strftime("%Y-%m-%d %H:%M:%S")
print(string_date)  # "2026-10-28 19:28:45"

# Convert String to Datetime object
date_input = "25/12/2026"
parsed_date = datetime.datetime.strptime(date_input, "%d/%m/%Y")
```

```python
Handy Formatting Directives

• %Y: 4-digit year (e.g., 2026)
• %m: 2-digit month (01 to 12)
• %d: 2-digit day of the month (01 to 31)
• %H: Hour in 24-hour format (00 to 23)
• %M: Minute (00 to 59)
• %A: Full weekday name (e.g., Wednesday)
```
