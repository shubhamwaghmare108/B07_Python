```python
certification = {"Dhurandhar": "A", "Bal Ganesh": "U"}

# Create indexed movie list
movie_list = list(certification.keys())

# Display menu cleanly
print("Available Movies:")
for i, title in enumerate(movie_list, start=1):
    print(f"{i}. {title}")

# User inputs
movie_num = int(input("Enter movie number: "))

if 1 <= movie_num <= len(movie_list):
    selected_movie = movie_list[movie_num - 1]
    grade = certification[selected_movie]
    
    age = int(input("Enter your age: "))

    # Simplified condition logic
    if grade == "U" or (grade == "A" and age >= 18):
        print("Ticket Booked Successfully")
    else:
        print("Your age group is restricted for this movie")
else:
    print("Invalid movie selection")
```
