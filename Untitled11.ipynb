```python
replay = "No"
counter = 0
while replay == 'No':
    try:
        counter += 1
        print(f"{counter}".center(20))
        print("Sorry Bro !!")
        replay = input("Did you forgive me ?\n'Yes' or 'No'").strip().title()
        if replay not in ['Yes','No']:
            raise ValueError("Invalied Replay")
    except ValueError:
        print("Try Replying Again")
        replay = input("Did you forgive me ?\n'Yes' or 'No'").strip().title()
else:
    print(f"Finally you have forgiven me after seeking forgiveness for {counter} times.\nThank God")
```
