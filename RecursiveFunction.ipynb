```python
def greet():
    print("Hello")
    greet()

greet()
```

```python
Important: Every recursive function needs a stopping condition to prevent infinite recursion.
```

```python
2. Two Essential Components of Recursion
A recursive function generally contains two important parts.
A. Base Case
The base case defines when the recursion should stop.
B. Recursive Case
The recursive case is where the function calls itself with a smaller or def count(n):
    if n == 0:          # Base case
        return

    print(n)
    count(n - 1)        # Recursive case

count(5)simpler problem.
```

```python
def count(n):
    if n == 0:          # Base case
        return

    print(n)
    count(n - 1)        # Recursive case

count(5)
```

```python
3. Understanding the Call Stack
Python uses a call stack to manage function calls.

When a function calls another function, Python remembers where the current function paused so that it can resume later.
```

```python
def show(n):
    if n == 0:
        return

    print("Entering:", n)
    show(n - 1)
    print("Exiting:", n)

show(3)
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)


print(factorial(5))
```

```python
def factorial(n):
    if not isinstance(n, int) or isinstance(n, bool) or n < 0:
        raise ValueError("n must be a non-negative integer")

    if n <= 1:
        return 1

    return n * factorial(n - 1)
```

```python
def natural_sum(n):
    if n == 0:
        return 0

    return n + natural_sum(n - 1)


print(natural_sum(5))
```

| Feature | Recursion | Iteration |
|---|---|---|
| Mechanism | Function calls itself | Uses loops |
| Stopping condition | Base case | Loop condition |
| Memory usage | Usually higher due to call stack | Usually lower |
| Readability | Often natural for recursive problems | Often simpler for repetitive tasks |
| Performance | Function-call overhead | Usually faster in Python |
| Risk | `RecursionError` for excessive depth | Infinite loop if condition never changes |
| Common uses | Trees, backtracking, divide-and-conquer | Counting, summation, repeated processing |

```python
Common Mistakes in Recursion
Mistake 1: Missing the base case
Mistake 2: Not moving toward the base case
Mistake 3: Forgetting return
Mistake 4: Confusing print() with return
```

```python
Calculate the product of numbers from 1 to N.
```

```python

def product(n):
    if n == 1:       # Base case
        return 1

    return n * product(n - 1)  # Recursive case


print(product(5))
```
