```python
price = []
quantity = []
total = 0
user_input = 0
```

```python
while user_input == 0:
    
    p = input("Enter Price").strip()
    q = input("Enter Quantity").strip()
    if p.isnumeric() and q.isnumeric():
        price.append(float(p))
        quantity.append(int(q))
    else:
        raise ValueError("Invalied Data")
    user_input = int(input("Enter '1' if done else '0'"))
```

```python
price
```

```python
quantity
```

```python
for x,y in zip(price,quantity):
    total += x * y
```

```python
total
```

```python
if total > 5000:
    discount = total * 0.1
    net_bill = total * 0.9
else:
    discount = 0
    net_bill = total
```

```python
print("Bill Summary".center(30,"*"))
print("Total Amount",total)
print("Discount",discount)
print("Payable Amount",net_bill)
print("Thank you for shopping with us")
print("End".center(30,"*"))
```

```python
A person enters expenses until they type done. 
Calculate total spending, average expense, highest expense, and the number of expenses above ₹500.
```

```python
user_input = input("Enter your expenses comma seperated if done then type done at last").split(",")
```

```python
user_input
```

```python
if user_input[-1].lower().strip() == 'done':
   expenses = user_input[0:-1] 
```

```python
expenses
```

```python
expenses = [float(i) for i in expenses]
```

```python
expenses
```

```python
total_spending = sum(expenses)
avg_spending = total_spending / len(expenses)
highest_spend = max(expenses)

above_500 = 0

for i in expenses:
    if i > 500:
      above_500 += 1 
```
