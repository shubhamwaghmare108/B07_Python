```python
try:
    while True:
        age = int(input("enter your age "))
        if age < 0 or age > 120:
            age = int(input("enter your age "))
        break   
except ValueError:
    print("Its not possible to convert given data into int")
except InvaliedAgeError:
    while True:
        age = int(input("enter your age "))
        if age < 0 or age > 120:
        raise InvaliedAgeError("Not a valied age ,please enter proper data")
        
```
